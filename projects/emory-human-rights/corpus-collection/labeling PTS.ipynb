```python
import pandas as pd
```

```python
df1 = pd.read_csv('human_rights_reports_1999-2010.csv')[['year', 'cow_code', 'section_title', 'section_subtitle', 'text']]
df2 = pd.read_csv('human_rights_reports_2011-2015.csv')[['year', 'cow_code', 'section_title', 'section_subtitle', 'text']]
df3 = pd.read_csv('human_rights_reports_2016-2023.csv')[['year', 'COW', 'section_title', 'section_subtitle', 'text']].rename(columns={'COW': 'cow_code'})
```

```python
combined = pd.concat([df1, df2, df3], ignore_index=True)
```

```python
pts = pd.read_csv('PTS-2024.csv', encoding='latin1')[['Year', 'COW_Code_A', 'PTS_A', 'PTS_H', 'PTS_S']]
```

```python
merged = pd.merge(
    combined,
    pts,
    how='left',
    left_on=['year', 'cow_code'],
    right_on=['Year', 'COW_Code_A']
).drop(columns=['Year', 'COW_Code_A'])
```

```python
merged.to_csv('human_rights_reports_labeled.csv', index=False)
```
