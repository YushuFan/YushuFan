```python
# Loading packages

import pandas as pd
import spacy
nlp = spacy.load("en_core_web_lg")
```

```python
# Loading data

before2017 = pd.read_csv("original files/before2017_fixed_wo_traintest.csv", encoding="ISO-8859-1")
master_US = pd.read_csv("original files/master_US_2021-2022.csv", encoding="ISO-8859-1")
```

```python
# Function removing stop words and named entities
def process_text(text):
    if pd.isna(text):
        return ""
    
    doc = nlp(text)
    
    filtered_tokens = [
        token.text for token in doc if not token.is_stop and token.ent_type_ == ""
    ]
    
    return " ".join(filtered_tokens)
```

```python
before2017["processed_texts"] = before2017["texts"].apply(process_text)
before2017.to_csv("NLP_before2017.csv", index=False)
```

```python
master_US["processed_texts"] = master_US["texts"].apply(process_text)
master_US.to_csv("NLP_master_US.csv", index=False)
```
